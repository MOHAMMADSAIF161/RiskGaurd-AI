```python
import pandas as pd
import numpy as np

# Reload dataset
DATA_PATH = "/content/RiskGuard_Return_Data/ecommerce_sales_34500.csv"

df = pd.read_csv(DATA_PATH)

# Convert fields
df["order_date"] = pd.to_datetime(df["order_date"])
df["returned_flag"] = (df["returned"] == "Yes").astype(int)

# Sort chronologically
df = df.sort_values(
    ["customer_id", "order_date", "order_id"]
).reset_index(drop=True)

# -----------------------------
# Previous customer history
# -----------------------------

g = df.groupby("customer_id", sort=False)

# Number of previous orders
df["previous_orders"] = g.cumcount()

# Previous returns
df["previous_returns"] = (
    g["returned_flag"]
    .cumsum()
    .shift(1)
    .fillna(0)
)

# IMPORTANT:
# Fix shift across customers
df["previous_returns"] = (
    df["previous_returns"]
    .where(
        df["customer_id"].eq(df["customer_id"].shift()),
        0
    )
)

# Previous return rate
df["previous_return_rate"] = np.where(
    df["previous_orders"] > 0,
    df["previous_returns"] / df["previous_orders"],
    0.0
)

# Previous spending
df["previous_total_spend"] = (
    g["total_amount"]
    .cumsum()
    .shift(1)
    .fillna(0)
)

df["previous_total_spend"] = (
    df["previous_total_spend"]
    .where(
        df["customer_id"].eq(df["customer_id"].shift()),
        0
    )
)

# Previous average order value
df["previous_avg_order_value"] = np.where(
    df["previous_orders"] > 0,
    df["previous_total_spend"] / df["previous_orders"],
    0.0
)

# Previous returned amount
df["previous_returned_amount"] = (
    (df["total_amount"] * df["returned_flag"])
    .groupby(df["customer_id"])
    .cumsum()
    .shift(1)
    .fillna(0)
)

df["previous_returned_amount"] = (
    df["previous_returned_amount"]
    .where(
        df["customer_id"].eq(df["customer_id"].shift()),
        0
    )
)

# Days since previous order
df["days_since_previous_order"] = (
    g["order_date"]
    .diff()
    .dt.total_seconds()
    .div(86400)
    .fillna(-1)
)

# -----------------------------
# Current-order derived features
# -----------------------------

df["discount_rate"] = np.where(
    df["price"] > 0,
    df["discount"] / df["price"],
    0
)

df["total_quantity_value"] = (
    df["price"] * df["quantity"]
)

# -----------------------------
# Verify
# -----------------------------

history_features = [
    "previous_orders",
    "previous_returns",
    "previous_return_rate",
    "previous_total_spend",
    "previous_avg_order_value",
    "previous_returned_amount",
    "days_since_previous_order"
]

print("Created features:")
print(history_features)

print("\nDataset shape:", df.shape)

print("\nFirst customer history:")
print(
    df[
        [
            "customer_id",
            "order_date",
            "returned",
            "previous_orders",
            "previous_returns",
            "previous_return_rate"
        ]
    ].head(15)
)
```

```python
import pandas as pd
import numpy as np

from xgboost import XGBClassifier
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# -----------------------------
# Target
# -----------------------------

y_train = (train_df["returned"] == "Yes").astype(int)
y_val   = (val_df["returned"] == "Yes").astype(int)
y_test  = (test_df["returned"] == "Yes").astype(int)

# -----------------------------
# Model features
# -----------------------------

NUMERIC_FEATURES = [
    "price",
    "discount",
    "quantity",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "delivery_time_days",
    "customer_age",

    # Customer history
    "previous_orders",
    "previous_returns",
    "previous_return_rate",
    "previous_total_spend",
    "previous_avg_order_value",
    "previous_returned_amount",
    "days_since_previous_order",

    # Derived
    "discount_rate",
    "total_quantity_value",
]

CATEGORICAL_FEATURES = [
    "category",
    "payment_method",
    "region",
    "customer_gender",
]

FEATURES = NUMERIC_FEATURES + CATEGORICAL_FEATURES

print("Number of features:", len(FEATURES))
print(FEATURES)
```

```python
# One-hot encode categorical columns
X_all = pd.get_dummies(
    df[FEATURES],
    columns=CATEGORICAL_FEATURES,
    dtype=int
)

# Re-create split indices using order_id
train_ids = set(train_df["order_id"])
val_ids = set(val_df["order_id"])
test_ids = set(test_df["order_id"])

X_train = X_all[df["order_id"].isin(train_ids)].copy()
X_val   = X_all[df["order_id"].isin(val_ids)].copy()
X_test  = X_all[df["order_id"].isin(test_ids)].copy()

# Make absolutely sure row counts align
assert len(X_train) == len(y_train)
assert len(X_val) == len(y_val)
assert len(X_test) == len(y_test)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

```python
from xgboost import XGBClassifier

negative = (y_train == 0).sum()
positive = (y_train == 1).sum()

scale_pos_weight = negative / positive

print("Negative:", negative)
print("Positive:", positive)
print("Scale pos weight:", round(scale_pos_weight, 4))

return_model_v1 = XGBClassifier(
    n_estimators=500,
    max_depth=5,
    learning_rate=0.05,
    subsample=0.85,
    colsample_bytree=0.85,
    objective="binary:logistic",
    eval_metric="aucpr",
    scale_pos_weight=scale_pos_weight,
    random_state=42,
    n_jobs=-1
)

return_model_v1.fit(
    X_train,
    y_train,
    eval_set=[(X_val, y_val)],
    verbose=False
)

print("\nReturn-Risk Model V1 trained successfully.")
```

```python
from sklearn.metrics import roc_auc_score, average_precision_score

val_prob = return_model_v1.predict_proba(X_val)[:, 1]

roc_auc = roc_auc_score(y_val, val_prob)
pr_auc = average_precision_score(y_val, val_prob)

print("Validation ROC-AUC:", round(roc_auc, 4))
print("Validation PR-AUC:", round(pr_auc, 4))
```

```python
# ============================================
# RiskGuard Return-Risk Model V2 Features
# ============================================

import numpy as np
import pandas as pd

# Work from the already chronologically sorted dataframe
df = df.sort_values(
    ["customer_id", "order_date", "order_id"]
).reset_index(drop=True)

g = df.groupby("customer_id", sort=False)

# -----------------------------
# 1. Previous average quantity
# -----------------------------

previous_quantity_sum = (
    g["quantity"]
    .cumsum()
    .shift(1)
    .fillna(0)
)

previous_quantity_sum = previous_quantity_sum.where(
    df["customer_id"].eq(df["customer_id"].shift()),
    0
)

df["previous_avg_quantity"] = np.where(
    df["previous_orders"] > 0,
    previous_quantity_sum / df["previous_orders"],
    0
)

# -----------------------------
# 2. Previous average discount
# -----------------------------

discount_sum = (
    g["discount"]
    .cumsum()
    .shift(1)
    .fillna(0)
)

discount_sum = discount_sum.where(
    df["customer_id"].eq(df["customer_id"].shift()),
    0
)

df["previous_avg_discount"] = np.where(
    df["previous_orders"] > 0,
    discount_sum / df["previous_orders"],
    0
)

# -----------------------------
# 3. Previous average delivery time
# -----------------------------

delivery_sum = (
    g["delivery_time_days"]
    .cumsum()
    .shift(1)
    .fillna(0)
)

delivery_sum = delivery_sum.where(
    df["customer_id"].eq(df["customer_id"].shift()),
    0
)

df["previous_avg_delivery_days"] = np.where(
    df["previous_orders"] > 0,
    delivery_sum / df["previous_orders"],
    0
)

# -----------------------------
# 4. Current amount vs history
# -----------------------------

df["amount_vs_customer_average"] = np.where(
    df["previous_avg_order_value"] > 0,
    df["total_amount"] / df["previous_avg_order_value"],
    1.0
)

# Cap extreme ratios
df["amount_vs_customer_average"] = (
    df["amount_vs_customer_average"]
    .clip(0, 20)
)

# -----------------------------
# 5. Customer return behavior
# -----------------------------

df["has_return_history"] = (
    df["previous_returns"] > 0
).astype(int)

df["repeat_customer"] = (
    df["previous_orders"] > 0
).astype(int)

# -----------------------------
# 6. Customer spending intensity
# -----------------------------

df["avg_spend_per_previous_order"] = np.where(
    df["previous_orders"] > 0,
    df["previous_total_spend"] / df["previous_orders"],
    0
)

# -----------------------------
# 7. Order economics
# -----------------------------

df["estimated_profit_value"] = (
    df["total_amount"] * df["profit_margin"]
)

df["shipping_to_order_ratio"] = np.where(
    df["total_amount"] > 0,
    df["shipping_cost"] / df["total_amount"],
    0
)

# -----------------------------
# 8. Quantity / order value
# -----------------------------

df["amount_per_item"] = np.where(
    df["quantity"] > 0,
    df["total_amount"] / df["quantity"],
    df["total_amount"]
)

# -----------------------------
# 9. Order date features
# -----------------------------

df["order_month"] = df["order_date"].dt.month
df["order_day_of_week"] = df["order_date"].dt.dayofweek

# -----------------------------
# Final V2 feature set
# -----------------------------

V2_NUMERIC_FEATURES = [
    "price",
    "discount",
    "quantity",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "delivery_time_days",
    "customer_age",

    # Existing customer history
    "previous_orders",
    "previous_returns",
    "previous_return_rate",
    "previous_total_spend",
    "previous_avg_order_value",
    "previous_returned_amount",
    "days_since_previous_order",

    # New behavioral features
    "previous_avg_quantity",
    "previous_avg_discount",
    "previous_avg_delivery_days",
    "amount_vs_customer_average",
    "has_return_history",
    "repeat_customer",
    "avg_spend_per_previous_order",

    # Order economics
    "estimated_profit_value",
    "shipping_to_order_ratio",
    "amount_per_item",

    # Date
    "order_month",
    "order_day_of_week",
]

V2_CATEGORICAL_FEATURES = [
    "category",
    "payment_method",
    "region",
    "customer_gender",
]

V2_FEATURES = (
    V2_NUMERIC_FEATURES +
    V2_CATEGORICAL_FEATURES
)

print("V2 logical features:", len(V2_FEATURES))
print(V2_FEATURES)
```

```python
# ============================================
# Prepare V2 datasets
# ============================================

# Make sure chronological ordering is preserved
df = df.sort_values(
    ["order_date", "order_id"]
).reset_index(drop=True)

# Use the exact date boundaries established earlier
train_df = df[
    df["order_date"].dt.date < train_cutoff
].copy()

val_df = df[
    (df["order_date"].dt.date >= train_cutoff) &
    (df["order_date"].dt.date < val_cutoff)
].copy()

test_df = df[
    df["order_date"].dt.date >= val_cutoff
].copy()

# Target
y_train_v2 = train_df["returned_flag"].astype(int)
y_val_v2 = val_df["returned_flag"].astype(int)
y_test_v2 = test_df["returned_flag"].astype(int)

# One-hot encoding
combined = pd.concat(
    [
        train_df[V2_FEATURES],
        val_df[V2_FEATURES],
        test_df[V2_FEATURES]
    ],
    axis=0
)

combined_encoded = pd.get_dummies(
    combined,
    columns=V2_CATEGORICAL_FEATURES,
    dtype=int
)

# Split back
n_train = len(train_df)
n_val = len(val_df)

X_train_v2 = combined_encoded.iloc[:n_train].copy()

X_val_v2 = combined_encoded.iloc[
    n_train:n_train + n_val
].copy()

X_test_v2 = combined_encoded.iloc[
    n_train + n_val:
].copy()

# Safety checks
assert len(X_train_v2) == len(y_train_v2)
assert len(X_val_v2) == len(y_val_v2)
assert len(X_test_v2) == len(y_test_v2)

print("V2 dataset prepared")
print("--------------------")
print("Logical features:", len(V2_FEATURES))
print("Encoded features:", X_train_v2.shape[1])
print("Train:", X_train_v2.shape)
print("Validation:", X_val_v2.shape)
print("Test:", X_test_v2.shape)
```

```python
from xgboost import XGBClassifier

# Class imbalance
negative_v2 = (y_train_v2 == 0).sum()
positive_v2 = (y_train_v2 == 1).sum()

scale_pos_weight_v2 = negative_v2 / positive_v2

print("Negative:", negative_v2)
print("Positive:", positive_v2)
print("Scale pos weight:", round(scale_pos_weight_v2, 4))

# Model V2
return_model_v2 = XGBClassifier(
    n_estimators=700,
    max_depth=4,
    learning_rate=0.03,
    min_child_weight=5,
    subsample=0.85,
    colsample_bytree=0.85,
    objective="binary:logistic",
    eval_metric="aucpr",
    scale_pos_weight=scale_pos_weight_v2,
    random_state=42,
    n_jobs=-1
)

return_model_v2.fit(
    X_train_v2,
    y_train_v2,
    eval_set=[(X_val_v2, y_val_v2)],
    verbose=False
)

print("\nReturn-Risk Model V2 trained successfully.")
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, average_precision_score

BASE_NUMERIC = [
    "price",
    "discount",
    "quantity",
    "total_amount",
    "shipping_cost",
    "profit_margin",
    "delivery_time_days",
    "customer_age",
    "previous_orders",
    "previous_returns",
    "previous_return_rate",
    "previous_avg_order_value",
    "days_since_previous_order",
]

BASE_CATEGORICAL = [
    "category",
    "payment_method",
    "region",
    "customer_gender",
]

X_train_base = train_df[BASE_NUMERIC + BASE_CATEGORICAL]
X_val_base = val_df[BASE_NUMERIC + BASE_CATEGORICAL]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            SimpleImputer(strategy="median"),
            BASE_NUMERIC
        ),
        (
            "cat",
            Pipeline([
                ("imputer", SimpleImputer(strategy="most_frequent")),
                (
                    "onehot",
                    OneHotEncoder(
                        handle_unknown="ignore"
                    )
                )
            ]),
            BASE_CATEGORICAL
        )
    ]
)

baseline_model = Pipeline([
    ("preprocessor", preprocessor),
    (
        "model",
        RandomForestClassifier(
            n_estimators=400,
            max_depth=8,
            min_samples_leaf=10,
            class_weight="balanced",
            random_state=42,
            n_jobs=-1
        )
    )
])

baseline_model.fit(
    X_train_base,
    y_train_v2
)

base_val_prob = baseline_model.predict_proba(
    X_val_base
)[:, 1]

base_roc_auc = roc_auc_score(
    y_val_v2,
    base_val_prob
)

base_pr_auc = average_precision_score(
    y_val_v2,
    base_val_prob
)

print("Baseline Random Forest")
print("----------------------")
print("ROC-AUC:", round(base_roc_auc, 4))
print("PR-AUC :", round(base_pr_auc, 4))
```
